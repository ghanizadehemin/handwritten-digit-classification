```python
import numpy as np
from sklearn import linear_model
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
digits=datasets.load_digits()
```

```python
digits.data.shape
```

```python
digits.target[1796]
```

```python
plt.imshow(digits.images[1796])
```

```python
x=digits.data
y=digits.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, random_state=42,stratify=y, test_size=0.2)
```

```python
model=linear_model.LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train, y_train)
```

```python
out=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

# مقایسه جواب‌های واقعی با پیش‌بینی‌های مدل
real_accuracy = accuracy_score(y_test, out)
print(f"دقت واقعی مدل: {real_accuracy * 100:.2f}%")
```

```python
right = (y_test == out).sum()
wrong = (y_test != out).sum()

print("Correct:", right)
print("Wrong:", wrong)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

# رسم ماتریس درهم‌ریختگی
ConfusionMatrixDisplay.from_predictions(y_test, out, cmap='Blues')
plt.title("Confusion Matrix - Handwritten Digit Classification")
plt.show()
```
